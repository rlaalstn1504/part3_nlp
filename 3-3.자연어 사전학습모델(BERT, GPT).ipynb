## 1. 사전학습 모델 소개

Transformer의 등장은 NLP에 큰 혁신을 가져왔습니다.  
하지만 Transformer 구조만으로는 모델을 **처음부터 매번 학습**해야 하는 부담이 있었습니다.  
이를 해결하기 위해 등장한 것이 **사전학습 모델(Pre-trained Model)** 입니다.  

- 대규모 말뭉치로 **일반 언어 패턴을 먼저 학습**  
- 이후 특정 작업에 맞게 **미세조정(Fine-tuning)**  
- 대표적 모델: **BERT (이해 중심)**, **GPT (생성 중심)**

## 2. 사전학습(Pre-training) 개념

### 2.1 사전학습이란?
**정의**: 대규모 텍스트 데이터셋을 활용해 **언어의 일반적 패턴(문법, 어휘, 의미 관계 등)** 을 먼저 학습한 뒤, 다양한 작업에 재사용하는 방식  

**절차**: 대규모 말뭉치로 언어의 일반적 특성을 학습한 뒤, 특정 작업(Task)에 맞게 모델을 세부 조정(Fine-tuning) 하는 2단계 학습 전략

**비유**  
- **사전학습**: 국어사전을 통째로 읽으며 기본 문법과 단어 습득  
- **미세조정**: 이후 토익/토플 같은 특정 시험에 맞춘 공부 

### 2.2 사전학습의 진행 방식

BERT나 GPT와 같은 모델은 **사전학습 단계에서 언어의 일반적인 규칙**을 익히도록 특별한 과제를 수행합니다.

대표적인 방법은:  
1. **빈칸 채우기(Masked Language Modeling, MLM)**  
   - 문장에서 단어 일부를 가리고(MASK) 모델이 맞히도록 학습  
   - 예: `"나는 [MASK]에 갔다." → "학교"`  

2. **다음 단어 예측(Language Modeling, LM)**  
   - 앞 단어들을 보고 다음에 올 단어를 예측  
   - 예: `"오늘은 날씨가" → "좋다"`  

3. **문장 관계 학습(Next Sentence Prediction, NSP, BERT 전용)**  
   - 두 문장이 실제로 이어지는지 아닌지 판별  
   - 예: `"나는 책을 읽었다. [SEP] 그래서 서점에 갔다." → 이어짐`  

이렇게 학습하면 모델은 **문장 구조, 단어 의미, 문맥 관계**를 일반적으로 이해할 수 있게 됩니다.  
자세한 방식은 BERT와 GPT 단원에서 구체적으로 다룹니다. 

### 2.3 실제 데이터 출처 (구글 사례)
대표적으로 **BERT(2018)** 모델은 다음 데이터셋으로 학습되었습니다:
- **Wikipedia** (영문 전체 문서 약 25억 단어)  
- **BookCorpus** (약 8억 단어 규모의 온라인 영어 책 데이터)  

GPT 계열 모델도 유사하게 인터넷의 **웹 텍스트, 뉴스, 위키, 공개 도메인 책** 등을 활용했습니다. 

### 2.4 장점
- **소량 데이터로도 높은 성능**: 특정 태스크에 맞게 소량 데이터만 추가 학습하면 성능 향상  
- **범용성**: 분류, 요약, 생성, 질의응답 등 다양한 태스크에 활용 가능  
- **개발 효율성**: 연구자/개발자가 처음부터 모델을 학습할 필요 없이, 이미 학습된 모델을 가져와 바로 사용 가능  

### 2.5 단점
- **데이터 편향(Bias)**: 위키, 책, 인터넷 데이터는 특정 문화·언어권 중심 → 모델도 편향된 결과 출력 가능  
- **자원 소모**: 수억~수천억 단어를 학습하려면 고성능 GPU/TPU, 막대한 전력 필요  
- **실시간 최신성 부족**: 학습 당시 데이터에 한정 → 최신 사건, 신조어 반영 어려움  
- **모델 크기**: 파라미터 수가 수억~수천억 개에 달해 저장/배포/서빙에 부담 

## 3. BERT (Bidirectional Encoder Representations from Transformers)
[BERT 논문](https://arxiv.org/pdf/1810.04805)

### 3.1 등장 배경
Transformer Encoder 구조를 활용하여, 문맥을 **양방향**으로 학습하는 모델  
→ 기존 언어 모델이 갖던 "한 방향 이해"의 한계를 극복  

### 3.2 학습 방식

#### 사전학습 종류

- **MLM (Masked Language Model)**  
  일부 단어를 [MASK]로 가리고 맞히게 학습  
  예: "나는 [MASK]에 갔다." → "학교"  
- **NSP (Next Sentence Prediction)**  
  두 문장이 이어지는 관계인지 판별  

#### Masked Learning이 학습하는 언어적 지식 유형
"Masked learning"이라 불리는 **빈칸 채우기 학습**을 반복하면서, 모델은 다양한 언어적 능력을 자연스럽게 습득하게 됩니다.  

| 작업(Task)           | 사전학습 중 학습할 수 있는 자연스러운 예시 문장 |
|----------------------|-----------------------------------------|
| 문법 (Grammar)       | 그는 어제 {도서관, 공원}에 갔다. |
| 어휘 의미 (Lexical Semantic) | 나는 아침에 {커피, 차}를 마셨다. |
| 세계 지식 (World Knowledge) | 미국의 수도는 {워싱턴, 뉴욕}이다. |
| 감정 분석 (Sentiment Analysis) | 리뷰: 이 드라마는 {재미있었다, 지루했다}. |
| 수학 문제 (Math Question) | 계산 문제: 5 × 6 = {30, 42} |
| 번역 (Translation)   | ‘고마워’를 영어로 하면 {thank you, hello}이다. |


따라서 BERT는 단순히 단어 맞추기를 넘어서,  
- 문법적 일관성  
- 단어 의미 파악  
- 상식/세계 지식  
- 감정적 뉘앙스  
- 간단한 수학 계산  
- 번역 패턴  

까지 간접적으로 배우게 됩니다.

### 3.3 BERT — 학습·추론 개요

BERT는 Transformer의 **Encoder 블록**을 기반으로 하며, **사전학습(Pre-training)** 단계와 **미세조정(Fine-tuning)**  
단계를 통해 문맥을 깊이 이해하는 언어 모델로 발전합니다.  

#### 입력 단계 (Input Representation)

BERT에 입력되는 문장은 단순한 텍스트가 아니라,  **세 가지 임베딩의 합**으로 구성된 벡터로 표현됩니다.

#### (1) 입력 구성 요소
<img src="image/bert_input.jpg" width="750">

- **Token Embedding**:  
  - 문장을 단어 또는 서브워드 단위로 분해하여 각 토큰을 고정된 벡터로 변환합니다.  
  - 예:  
    ```
    문장: [CLS] 그는 사과를 먹었다 [SEP]
    토큰화: ['[CLS]', '그', '##는', '사과', '##를', '먹었다', '[SEP]']
    ```

- **Segment Embedding**:  
  - BERT는 두 문장을 동시에 입력받을 수 있습니다(A문장과 B문장).  
  - 이때 문장 구분을 위해 **문장 A에는 0**, **문장 B에는 1**을 부여합니다.  
  - 예:  
    ```
    문장1: 나는 학교에 갔다.
    문장2: 수업이 재미있었다.
    → Segment A(0): [CLS] 나는 학교에 갔다 [SEP]
      Segment B(1): 수업이 재미있었다 [SEP]
    ```

- **Position Embedding**:  
  - Transformer에는 순서를 인식하는 능력이 없기 때문에,  
    단어의 **위치 정보를 인코딩한 벡터**를 추가합니다.  
  - 예:  
    ```
    [CLS] = 위치 0, 그는=1, 사과를=2, 먹었다=3, [SEP]=4
    ```

최종적으로, **입력 벡터는 Token + Segment + Position** 이 세 임베딩의 합으로 만들어집니다.

#### (2) 사전학습 단계 (Pre-training)

BERT는 대규모 말뭉치(위키백과, BookCorpus 등)에서  
아래 두 가지 **비지도 학습(Self-Supervised)** 과제를 통해 언어를 학습합니다.

#### Masked Language Modeling (MLM)
- 입력 문장에서 **15% 정도의 단어를 [MASK]로 가리고**,  
  그 단어를 예측하도록 학습합니다.
- 예:  
입력: 그는 [MASK]를 먹었다.
정답: 사과

- 이렇게 하면 모델이 **왼쪽과 오른쪽 문맥을 동시에 고려**해야 하기 때문에,  
**양방향(Self-Attention)** 학습이 가능해집니다.

#### `[MASK]` 토큰이란?
> 원래 단어를 임시로 가려서 모델이 스스로 문맥을 이용해 복원하도록 돕는 "빈칸 표시용" 토큰입니다.  
> Transformer의 Attention 구조 덕분에, 앞뒤 단어를 모두 참조하면서 문장을 이해할 수 있습니다.



#### Next Sentence Prediction (NSP)
- 두 문장을 입력받아, 두 번째 문장이 첫 번째 문장의 **자연스러운 다음 문장인지** 예측하는 과제입니다.  
- 예:   

| 문장 A | 문장 B | 라벨 | 
|---------|---------|-------|
| 나는 학교에 갔다. | 수업이 재미있었다. | True |
| 나는 학교에 갔다. | 위스키를 마셨다. | False |

- 이를 통해 BERT는 **문장 간 관계와 연결성**까지 학습하게 됩니다.

#### Transformer Encoder 내부 작동

사전학습 중 입력된 토큰 시퀀스는 **여러 개의 Transformer Encoder 블록**을 통과합니다.

<img src="image/tf_encoder.jpg">

각 블록은 다음으로 구성됩니다:
1. **Self-Attention Layer**  
 - 각 단어가 문장 내 모든 다른 단어와의 관계를 계산하여 문맥적 중요도를 반영합니다.  
 - 예: "그는" ↔ "먹었다" 관계를 이해  

2. **Feed Forward Neural Network (FFNN)**  
 - Self-Attention의 출력을 비선형 변환하여 더 복잡한 의미를 학습  

3. **Residual Connection + Layer Normalization**  
 - 입력 정보를 보존하면서 학습 안정성 향상  

> 이 과정을 여러 번 반복하며,  
> BERT는 단어 하나하나에 대해 문맥이 반영된 벡터(=Contextual Embedding)를 생성합니다

#### 출력 표현 (Contextual Embedding)

Encoder를 통과한 후, BERT는 각 토큰별로 문맥이 반영된 벡터를 출력합니다.

<img src="image/bert_finetuning.jpg" width="600">

- **[CLS] 토큰**  
- 문장 맨 앞에 항상 추가되는 특수 토큰으로, **문장 전체의 요약 의미를 대표**합니다.  
- Fine-tuning 시에는 분류기(Classifier) 층과 연결되어 감정 분석, 문서 분류 등에 활용됩니다.

- **[SEP] 토큰**  
- 문장 구분용 특수 토큰으로, 여러 문장을 하나의 입력으로 넣을 때 경계를 표시합니다.  

- **[MASK] 토큰**  
- MLM 학습에서 사용되며,  
  가려진 단어를 예측하는 훈련 신호로 작동합니다.

### 3.4 미세조정 단계 (Fine-tuning)

사전학습이 완료된 BERT는 다양한 NLP 작업에 맞게 **파인튜닝(Fine-tuning)** 됩니다.

- 사전학습된 파라미터를 초기값으로 사용  
- 작업별 특성에 따라 **얇은 출력층(Linear layer)** 만 추가  
- 전체 네트워크를 다운스트림 데이터에 맞게 미세하게 조정  

예시:  
| 작업 | 입력 | 출력 | 사용되는 벡터 |
|------|------|------|---------------|
| 문장 분류 | 한 문장 | 감정/주제 라벨 | `[CLS]` (마지막 층의 hidden state, Linear(768→num_labels)) |
| 질의응답(QA, SQuAD 등) | 질문 + 문단 | 정답 시작/끝 위치 | 각 단어별 Context 벡터 (Linear(768→2), start/end 예측) |
| 문장 유사도 | 두 문장 | 유사도 점수 | `[CLS]` 또는 모든 토큰 평균 벡터 (mean pooling, Linear(768→1) 또는 cosine) |

대부분의 작업은 **BERT의 인코더 구조를 그대로 유지**하고, 단 하나의 선형층을 추가해 학습합니다.

### 3.5 응용 분야
BERT는 **문맥 이해(understanding)** 능력에 특화된 모델로, 다음과 같은 다양한 작업에 활용됩니다:

- **문서 분류(Document Classification)**  
  뉴스 기사 카테고리 분류, 스팸 메일 탐지, 주제 분류 등  

- **감정 분석(Sentiment Analysis)**  
  영화 리뷰 긍/부정 판단, 고객 피드백 분석  

- **질의응답(Question Answering, QA)**  
  주어진 문단에서 질문에 해당하는 답 추출 (예: SQuAD 데이터셋)  

- **문장 유사도 측정(Sentence Similarity / Semantic Textual Similarity)**  
  두 문장의 의미가 얼마나 비슷한지 계산 (검색·추천 시스템 활용)  

- **개체명 인식(Named Entity Recognition, NER)**  
  문장에서 사람, 장소, 기관, 날짜 등 고유명사 추출  

- **의미역 결정(Semantic Role Labeling)**  
  문장에서 "누가 무엇을 하였는가"와 같은 역할 분석  

- **텍스트 요약(Text Summarization, Extractive 방식)**  
  문서에서 중요한 문장을 추출해 요약  

- **정보 검색(IR, Information Retrieval)**  
  검색 쿼리와 문서의 의미적 일치 여부 판단  

#### **요약**  

BERT는 입력 임베딩 구성 → 사전학습(MLM + NSP) → Transformer Encoder로 문맥 학습 → Fine-tuning으로  
실제 작업 적용의 과정을 통해,**언어 이해 능력을 사전 습득하고 다양한 NLP 과제에 재활용 가능한 범용 모델**이 되었습니다.

### 3.6 Hugging Face란?

#### 개요
**Hugging Face**는 인공지능(AI), 특히 **자연어 처리(NLP)** 모델을 쉽고 빠르게 활용할 수 있도록 만든  
오픈소스 AI 생태계 플랫폼입니다.

한마디로 말해, AI 모델의 GitHub + Colab + 커뮤니티를 결합한 공간이라 할 수 있습니다.  
전 세계 연구자와 개발자들이 사전학습된 모델, 데이터셋, 학습 코드 등을 공유하고 협업할 수 있는  
가장 대표적인 AI 허브입니다.

#### 모델과 토크나이저 불러오기

Hugging Face의 transformers 라이브러리에서 사전학습된 모델(AutoModel)과 토크나이저(AutoTokenizer)를 불러옵니다.

````python
import torch
# 사전학습된(pretrained) BERT 모델과 토크나이저를 쉽게 불러오는 클래스들을 가져옵니다.
from transformers import AutoModelForSequenceClassification, AutoTokenizer

# 사용할 사전학습 모델 이름 지정
# "bert-base-uncased"는 소문자 기반 영어 BERT 모델 (12-layer, 110M parameters)
model_name = "bert-base-uncased"

# 사전학습된 토크나이저 불러오기
# → 입력 문장을 WordPiece 단위로 쪼개어 숫자 ID 시퀀스로 변환함
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 사전학습된 모델 불러오기
# → 문장 분류(Sequence Classification)용 BERT 구조 로드
# from_pretrained()는 내부적으로 Hugging Face Hub에서 모델 가중치를 다운로드함
model = AutoModelForSequenceClassification.from_pretrained(model_name)
````

````python
# 예시 문장 하나 준비
text = "Hugging Face is creating a great ecosystem for NLP."

#  토크나이저로 텍스트 전처리
#  - 문장을 BERT가 이해할 수 있는 숫자 토큰(id)으로 변환
#  - return_tensors="pt" → PyTorch 텐서(torch.tensor) 형태로 반환
#  - 결과: {'input_ids': tensor(...), 'attention_mask': tensor(...)} 형태의 딕셔너리
inputs = tokenizer(text, return_tensors="pt")
````

````python
# 모델에 입력 전달 (추론 단계)
#  - **inputs 딕셔너리의 키**(input_ids, attention_mask 등)를 그대로 풀어서 전달(**inputs)
#  - 모델은 입력 문장을 인코딩하여 각 클래스(예: 긍정/부정)에 대한 점수를 계산
outputs = model(**inputs)
````

````python
# 출력 결과 확인
#  - outputs는 transformers의 SequenceClassifierOutput 객체
#  - outputs.logits는 (배치 크기, 클래스 개수) 형태의 점수(logit) 텐서
#  - 예: tensor([[-1.203, 2.315]]) → 두 클래스(NEGATIVE, POSITIVE)에 대한 점수
#  - softmax를 적용하면 각 클래스의 확률로 변환할 수 있음
print(outputs.logits)
````

````python
# pipeline()을 사용하면, 복잡한 전처리나 모델 로드를 몰라도 단 한 줄로 다양한 NLP 작업을 수행할 수 있습니다.

from transformers import pipeline

# 감정분석 예시
classifier = pipeline("sentiment-analysis")
result = classifier("I love using Hugging Face models!")
print(result)

# 출력 예시:
#[{'label': 'POSITIVE', 'score': 0.9998}] 
````

#### Fine-tuning을 위한 기본 구조

파인튜닝 시 많이 사용하는 구성은 **Trainer API**입니다.

````python
# transformers 라이브러리에서 파인튜닝(Fine-tuning)을 위한 주요 클래스 불러오기
#  - AutoModelForSequenceClassification : 문장 분류용 사전학습 모델
#  - Trainer / TrainingArguments : 학습 과정을 단순화시켜주는 고수준 API
#  - load_dataset : 공개 데이터셋을 쉽게 불러올 수 있는 함수 (datasets 라이브러리)
from transformers import AutoModelForSequenceClassification, AutoTokenizer, Trainer, TrainingArguments
from datasets import load_dataset


# 데이터 불러오기 -------------------------------------------------------------
# GLUE(SST-2) 데이터셋 로드
#  - GLUE : General Language Understanding Evaluation (언어 이해 벤치마크)
#  - SST-2 : Stanford Sentiment Treebank v2 (감정분석 데이터셋)
#  - 입력: 문장(sentence)
#  - 출력: 감정 라벨(label) → 0(부정) / 1(긍정)
dataset = load_dataset("glue", "sst2")

# dataset 구조 예시:
# dataset['train'][0]  → {'sentence': 'a stirring , funny and finally transporting re-imagining of beauty and the beast and 1930s horror films', 'label': 1}
````

````python
# 모델 및 토크나이저 불러오기 -----------------------------------------------
model_name = "bert-base-uncased"   # 영어 소문자 기반 BERT 모델 (12-layer, 110M parameters)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# AutoModelForSequenceClassification : 문장 분류 태스크용 BERT
# num_labels=2 → 긍정/부정 두 가지 감정 클래스를 예측하도록 출력층(Linear(768→2)) 추가
model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=2)
````

````python
# 데이터 전처리 (토크나이징)
# BERT는 문장을 단어가 아닌 WordPiece 단위로 쪼개서 입력받음
# truncation=True → 문장이 너무 길면 최대 길이(512토큰)까지만 자름
# padding=True → 배치(batch) 내 문장 길이를 동일하게 맞춤
def preprocess_function(examples):
    return tokenizer(examples["sentence"], truncation=True, padding=True)

# dataset.map()을 통해 전체 데이터셋에 전처리 함수를 적용
# batched=True → 여러 샘플을 한 번에 처리하여 속도 향상
encoded_dataset = dataset.map(preprocess_function, batched=True)

# encoded_dataset 예시:
# {'input_ids': [...], 'attention_mask': [...], 'label': ...}


# 학습 설정 (TrainingArguments)
training_args = TrainingArguments(
    output_dir="./results",          # 학습 결과(모델, 로그) 저장 경로
    eval_strategy="epoch",     # 한 epoch마다 검증(validation) 수행
    learning_rate=2e-5,              # 학습률 (BERT 권장값: 2e-5 ~ 5e-5)
    per_device_train_batch_size=32,  # GPU당 배치 크기
    num_train_epochs=2,              # 전체 데이터셋 반복 학습 횟수
    weight_decay=0.01,               # L2 정규화(가중치 감쇠) 계수 → 과적합 방지
)
from transformers import DataCollatorWithPadding

data_collator = DataCollatorWithPadding(tokenizer=tokenizer)

# Trainer 객체 생성 및 학습 시작 
# Trainer는 학습 루프, 손실 계산, 검증, 로깅 등을 자동으로 처리하는 high-level API
trainer = Trainer(
    model=model,                           # 학습할 모델
    args=training_args,                    # 학습 설정
    train_dataset=encoded_dataset["train"],# 학습 데이터셋
    eval_dataset=encoded_dataset["validation"], # 검증 데이터셋
    data_collator=data_collator
)

# 학습 시작 (fine-tuning)
# 내부적으로 optimizer, scheduler, evaluation loop까지 자동 실행
trainer.train()
````

````python
# 평가(Evaluation)
# 학습된 모델의 성능을 검증(validation set) 데이터로 평가
# Trainer.evaluate()는 자동으로 손실(loss) 및 정확도(metrics)를 계산
# (metrics를 별도로 설정하지 않으면 loss만 출력됨)

eval_results = trainer.evaluate()
print(" 평가 결과:")
print(eval_results)
# 출력 예시:
# {'eval_loss': 0.35, 'eval_runtime': 15.2, 'eval_samples_per_second': 200.5, ...}
````

````python
# 새로운 문장에 대한 추론(Inference)
# 학습된 모델을 이용해 감정(긍정/부정)을 예측

text = "The movie was surprisingly good and emotional."

# 입력 문장 토큰화 (모델 입력 형태로 변환)
inputs = tokenizer(text, return_tensors="pt")

# 모델을 평가 모드로 변경 (Dropout 등 비활성화)
model.eval()

# 추론 실행 (Gradient 계산 비활성화 → 더 빠르고 메모리 절약)
with torch.no_grad():
    outputs = model(**inputs)

# 모델 출력(logits)은 각 클래스(0=부정, 1=긍정)에 대한 점수값
logits = outputs.logits

# Softmax로 확률화 후 argmax로 예측 클래스 결정
predicted_class = torch.argmax(torch.softmax(logits, dim=1)).item()

# 결과 출력
labels = ["Negative", "Positive"]
print(f"\n📝 입력 문장: {text}")
print(f"🔮 예측 결과: {labels[predicted_class]} (확률: {torch.softmax(logits, dim=1).max().item():.4f})")
````

### 문제 3. `[CLS]` 토큰의 역할
BERT의 입력은 `[CLS] 문장A [SEP] 문장B [SEP]` 형태로 구성됩니다.  
이 중 `[CLS]` 토큰은 **사전학습(Pre-training)** 단계와 **파인튜닝(Fine-tuning)** 단계 모두에서 중요한 역할을 합니다.  

NSP 과제와 문장 분류 태스크에서 `[CLS]` 토큰의 역할을 설명하세요.

<details><summary>정답 보기</summary>

`[CLS]` 토큰은 BERT의 입력 맨 앞에 추가되어 **문장 전체를 대표하는 벡터 역할**을 합니다.  

NSP에서는 `[CLS]`의 **최종 hidden state(768차원)** 를 이용하여  
문장 A와 문장 B의 관계(연속된 문장인지, 무관한 문장인지)를 판단합니다.

- 입력: `[CLS] 문장A [SEP] 문장B [SEP]`  
- 출력: Linear(768 → 2) → Softmax → `[P(isNext), P(notNext)]`  
- 학습 라벨:  
  - 1 → 문장B는 문장A의 다음 문장  
  - 0 → 문장B는 랜덤하게 선택된 문장  

즉, `[CLS]` 벡터는 **"문장 쌍 관계 분류기(NSP head)"의 입력**으로 사용됩니다.


#### 파인튜닝(Fine-tuning) 단계
사전학습을 마친 후 `[CLS]` 벡터는  
문장 전체 의미를 대표하는 벡터로 간주되어  
**감정분석, 문장분류, 문장유사도 등 다양한 태스크**의 입력으로 활용됩니다.

- 예: 감정분류 시  
  `[CLS]` → Linear(768 → num_labels) → Softmax

#### 요약
| 구분 | 단계 | 역할 | 출력층 구조 |
|------|------|------|-------------|
| NSP | 사전학습 | 문장A-B의 관계 판단 | Linear(768 → 2) |
| 문장분류 | 파인튜닝 | 문장 전체 의미 벡터 | Linear(768 → num_labels) |


**결론:**  
> `[CLS]` 토큰은 BERT가 문장 전체를 하나의 벡터로 표현하기 위한 특수 토큰이며,  
> 사전학습(NSP)에서는 문장 관계 분류 입력으로,  
> 파인튜닝에서는 문장 전체 의미를 대표하는 특징 벡터로 사용된다.

### 문제 5. WordPiece 토크나이징의 원리와 필요성  
BERT는 입력 문장을 처리하기 위해 **WordPiece 토크나이저**를 사용합니다.  
다음 설명을 참고하여, WordPiece가 사용되는 이유와 동작 방식을 서술하세요.  

(힌트: 희귀 단어나 신조어, 접사, 철자 변형 단어도 처리 가능해야 합니다.)

<details><summary>정답 보기</summary>

WordPiece는 **BERT에서 사용하는 서브워드(Subword) 기반 토크나이징 방식**으로,  
전체 어휘를 단어 단위로 저장하지 않고 **단어를 더 작은 의미 단위로 분리**합니다.  

#### WordPiece가 필요한 이유
- 단어 단위 토크나이저(Word-level tokenizer)는  
  신조어나 철자 변형(예: "unknowingly", "micro-transaction")을 처리하지 못하고  
  `[UNK]`(unknown) 토큰으로 대체해버립니다.  
- 반면 WordPiece는 **자주 등장하는 단어는 그대로**,  
  **희귀 단어는 여러 조각(subword)** 으로 분리하여 표현할 수 있습니다.

예시:
```
Input:  "unbelievable"
Tokenized: ["un", "##bel", "##ievable"]
```
- "##" 표시는 "이 조각이 앞 단어 뒤에 이어진(subword)" 것을 의미  
- 자주 쓰이는 "un", "believe", "able" 조각으로 분리되어 `[UNK]` 발생을 줄임

#### WordPiece의 작동 방식
1. **초기 어휘(vocabulary)**: 문자 단위에서 시작 (a, b, c, ...).  
2. **병합 규칙 학습(merge rules)**:  
   - 코퍼스에서 자주 함께 등장하는 문자쌍을 병합하여 새로운 subword 생성  
   - 예: "th" → "the" → "there" → "therefore" 순으로 어휘 확장  
3. **최종 어휘 집합 생성**:  
   - 자주 등장하는 단어는 그대로 유지  
   - 희귀 단어는 subword들의 조합으로 표현

#### BERT 입력에서의 결과
```
문장: "Playing footballers happily"
토큰: ["[CLS]", "play", "##ing", "football", "##ers", "happily", "[SEP]"]
```

### 문제 14. NSP의 입력 형식
BERT에서 NSP(Next Sentence Prediction)는 어떤 방식으로 두 문장을 구분하나요?

<details><summary>정답 보기</summary>

BERT의 NSP는 두 문장이 실제로 이어지는 문장인지 아닌지를 예측하는 과제입니다.  
이를 위해 BERT는 두 문장을 하나의 입력으로 합치되, 다음 두 가지 정보를 활용해 문장 경계를 명확히 구분합니다.  

#### **Special Token 기반 구분**

입력 형식은 다음과 같습니다:

```
[CLS] 문장A [SEP] 문장B [SEP]
```

- `[CLS]` : 전체 문장의 대표 벡터  
- 첫 번째 `[SEP]` : 문장 A의 끝  
- 두 번째 `[SEP]` : 문장 B의 끝  

이 구조만으로도 두 문장의 경계가 명확해집니다.


#### **Segment Embedding(Token Type Embedding) 사용**

BERT는 각 토큰에게  
**문장 A 또는 문장 B에 속하는지 표시하는 추가 임베딩**을 더합니다.

- 문장 A 토큰 → Segment ID = 0  
- 문장 B 토큰 → Segment ID = 1  

이 정보 덕분에 모델은  
"이 토큰이 어느 문장에 속하는가?"를 명확히 구분할 수 있습니다.


즉, NSP는 **special token으로 문장 구조를 표시하고,  
segment embedding으로 문장 소속을 구분하는 방식**으로 구현됩니다.
</details>

### 문제 15. Encoder만 사용한 이유
BERT가 Transformer의 Encoder만 사용하는 이유는?

<details><summary>정답 보기</summary>


BERT의 목적은 **문장을 깊고 양방향적으로 이해하는 것(Contextual Understanding)** 입니다.  
이 목표에 가장 적합한 구조가 바로 Transformer의 **Encoder**입니다.


#### Encoder는 Bidirectional Self-Attention 사용

Encoder는 문장의 모든 위치가 서로를 자유롭게 참고할 수 있어  
**왼쪽 → 오른쪽, 오른쪽 → 왼쪽**  
양방향 문맥을 모두 활용할 수 있습니다.

예를 들어 "bank"라는 단어를 이해할 때  
앞과 뒤의 단어를 모두 고려할 수 있습니다.


#### Decoder는 생성(generation) 목적이며 미래 단어를 보지 못하도록 설계됨

Decoder는 autoregressive 구조라 현재 단어를 예측할 때 미래 단어를 보면 안 됩니다(masking).

따라서 Decoder는 본질적으로 **양방향 문맥 이해보다는 텍스트 생성에 최적화된 구조**입니다.
</details>

### 문제 16. BERT에서 cased와 uncased 모델의 차이
BERT 모델에서 **cased**와 **uncased**의 차이는 무엇이며, 각각 언제 사용하는 것이 적절한가?

<details><summary>정답 보기</summary>

BERT의 **cased / uncased** 차이는  
👉 **대소문자 정보를 유지하느냐, 제거하느냐**에 있습니다.


#### Uncased 모델: 대소문자 무시

- 모든 입력을 **소문자로 변환**하여 학습
- 대소문자 정보를 완전히 제거
- 예:
  - "Apple", "apple", "APPLE" → 모두 "apple"

**특징**
- 데이터가 적을 때 일반화 성능이 좋음
- 대소문자가 의미에 크게 영향을 주지 않는 작업에 적합

**주 사용처**
- 감성 분석
- 문장 분류
- 일반적인 의미 이해(Task-focused NLP)


#### Cased 모델: 대소문자 유지

- 입력 텍스트의 **대소문자를 그대로 유지**
- "Apple" ≠ "apple"

**특징**
- 대소문자가 의미 구분에 중요한 정보를 제공
- 모델이 더 많은 언어적 단서를 활용 가능

**주 사용처**
- 개체명 인식(NER)
- 고유명사 구분이 중요한 태스크
- 문서 분석, 정보 추출


#### 왜 두 가지 버전이 존재할까?

대소문자는 항상 도움이 되지는 않습니다.

- 불필요한 경우 → 노이즈가 될 수 있음
- 필요한 경우 → 중요한 의미 단서가 됨

그래서 BERT는  
**일반화에 강한 uncased**와  
**정밀한 의미 구분이 가능한 cased**  
두 가지 버전을 모두 제공합니다.


#### 한 줄 요약

- **Uncased**: 의미 중심, 안정적인 성능, 일반적인 NLP 작업
- **Cased**: 형태·고유명사 구분 중요, 정밀한 언어 이해
</details>


````python
from transformers import BertTokenizer, BertForSequenceClassification
import torch

# 1) 토크나이저 로드
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# 2) 문장을 숫자 ID로 변환
inputs = tokenizer("I love NLP", return_tensors="pt")

# 3) 사전학습된 BERT 분류 모델 불러오기
model = BertForSequenceClassification.from_pretrained("bert-base-uncased")

# 4) 추론 실행
outputs = model(**inputs)

# 5) 결과 확인 (클래스별 점수)
print(outputs.logits) 
````

### 4.1 등장 배경
BERT가 **양방향 인코더(Encoder)** 구조를 사용하여 문맥 이해에 강점을 보였다면, GPT는 **디코더(Decoder)**  
구조를 채택하여 **문장 생성(generation)** 에 특화된 모델입니다.

- BERT는 [MASK] 토큰을 예측하기 때문에, 문장 내의 **공백 채우기(masked prediction)** 에는 강하지만  
  다음 단어 이어쓰기(next word prediction)와 같은 자연스러운 생성에는 부적합합니다.  

- OpenAI는 **Transformer의 Decoder 부분만 활용**하여, 왼쪽에서 오른쪽으로 순차적으로  
 단어를 생성하는 **자기회귀 언어모델(Auto-regressive LM)** 구조를 설계했습니다.  

- GPT는 이런 방식으로 **생성적(Generative)** 언어모델의 새로운 패러다임을 열었으며, 이후 GPT-2, GPT-3,  
 GPT-4로 이어지며 대규모 파라미터 확장을 통해 **few-shot / zero-shot 학습 능력**을 보이는 초거대 모델로 발전했습니다.

### 4.2 GPT 구조 한눈에 보기 (Decoder-only LM)

- GPT는 **Transformer 인코더가 아니라 디코더 블록만 여러 층 쌓은 구조**입니다.

- 각 디코더 블록 안의 **Self-Attention에 Causal Mask(삼각 마스크)를 적용**해서 현재 위치의 토큰이 **오른쪽(미래) 토큰을 보지 못하도록** 막습니다.

- 그래서 모델은 항상 **왼쪽(이전 단어들)만 보고 다음 단어를 예측**하는 **단방향(auto-regressive) 언어 모델**이 됩니다.

<img src="image/GPT.png" width="500">


수식으로 쓰면, 문장 $x = (x_1, \dots, x_T)$에 대해 아래처럼, **각 시점 단어 $x_t$는 그 이전 단어들 $x_{<t}$만 보고 조건부 확률을 예측**하도록 학습됩니다.

$p(x) = p(x_1, \dots, x_T) = \prod_{t=1}^T p(x_t \mid x_{<t})$



### 4.3 GPT-1의 학습 방식 (Shape 기반 상세 설명)

GPT-1은 **Autoregressive Language Model**로, 이전까지의 단어 시퀀스를 보고 다음 단어를 예측하는 모델입니다.

GPT에서 사용하는 확률식은 다음과 같습니다:

- 여기서 **w는 word(또는 token)의 약자**입니다.

$$
P(w_t \mid w_1, w_2, \ldots, w_{t-1})
$$

즉, **이전 단어들을 기반으로 다음 단어를 확률적으로 예측하는 것**입니다.


#### 1) 입력 형식 (Embedding 단계)

GPT-1의 실제 설정은 다음과 같습니다:

- 최대 시퀀스 길이: 512  
- Hidden size: 768  
- Vocabulary size: 약 40,000  
- Layer 개수: 12  

입력 토큰 ID의 shape:

$$
(512,)
$$

토큰 임베딩 + 위치 임베딩을 더하면:

$$
X \in \mathbb{R}^{512 \times 768}
$$

이 행렬이 Transformer Decoder 블록으로 입력됩니다.

#### 2) Causal Mask (미래 정보 차단)

GPT-1은 다음 단어 예측을 위해 *미래 토큰을 볼 수 없도록* 해야 합니다.

<img src="image/casual_mask.png" width="500">

이를 위해 **하삼각형(Tril) 형태의 Causal Mask**를 사용합니다.

Mask의 shape:

$$
M \in \mathbb{R}^{512 \times 512}
$$

- **하삼각형(대각선 포함)**: 0 (참조 가능)  
- **상삼각형(미래 토큰)**: $-\infty$ (softmax 후 0으로 처리됨)



#### 3) Self-Attention의 Shape 변화

Q, K, V는 다음과 같이 계산됩니다:

$$
Q, K, V \in \mathbb{R}^{512 \times 768}
$$

Attention score 계산:

$$
QK^\top \in \mathbb{R}^{512 \times 512}
$$

Causal Mask 적용:

$$
\text{MaskedScores} = QK^\top + M
$$

Softmax 후 Attention weight:

$$
A \in \mathbb{R}^{512 \times 512}
$$

최종 Attention Output:

$$
AV \in \mathbb{R}^{512 \times 768}
$$

GPT-1의 12-layer Decoder 블록 전체를 통과해도 hidden state shape은 동일합니다.

$$
512 \times 768
$$





#### 4) LM Head (출력층)

Transformer의 마지막 hidden state:

$$
H \in \mathbb{R}^{512 \times 768}
$$

LM Head (Linear Layer) 파라미터:

$$
W \in \mathbb{R}^{768 \times 40000}
$$

출력 로짓(logits):

$$
\text{logits} = HW \in \mathbb{R}^{512 \times 40000}
$$

→ **각 위치(512개)에서 40,000개 단어에 대한 다음 단어 확률분포를 동시에 예측합니다.**



#### 5) 손실 함수 (Cross-Entropy Loss)

GPT-1은 각 시점 \(t\)에서 현재까지의 토큰을 기반으로  
다음 토큰 \(w_{t+1}\)을 예측하는 **Causal Language Modeling** 방식으로 학습됩니다.

각 시점 \(t\)에서 모델의 출력은 vocabulary 전체에 대한 logit 벡터입니다.

- logits at time \(t\):  
$$
\text{logits}_t \in \mathbb{R}^{|\mathcal{V}|} \quad (|\mathcal{V}| = 40{,}000)
$$

- target token:  
$$
w_{t+1}
$$

전체 시퀀스 길이가 \(T = 512\)일 때,  
총 \(T-1 = 511\)개의 next-token prediction에 대해 loss를 계산하며,  
sequence-level loss는 다음과 같이 정의됩니다.

$$
\mathcal{L}
= \frac{1}{T-1}
\sum_{t=1}^{T-1}
\mathrm{CE}(\text{logits}_t,\; w_{t+1})
$$

즉, GPT-1은 **한 번의 forward pass로 모든 시점의 next-token prediction을 동시에 계산**하며,  
이는 일부 토큰만 예측하는 BERT의 Masked Language Modeling(MLM) 방식과 대조됩니다.


### 4.4 GPT의 두 단계: 사전학습과 파인튜닝

GPT도 BERT와 마찬가지로 두 단계로 학습됩니다

1. **사전학습 (Pre-training)**  
   - 수백 GB~수 TB 규모의 대규모 텍스트 데이터(웹, 위키, 책, 뉴스 등)를 사용  
   - 인간이 별도로 라벨링하지 않은 **비지도(unlabeled)** 데이터 기반  
   - 목표: **일반적인 언어 패턴, 어휘, 문맥 구조**를 학습  
   - 학습 방법: "이전 단어를 보고 다음 단어 예측" (Auto-regressive LM)  

   예:  
   ```
   입력: "The Eiffel Tower is located in"
   출력: "Paris"
   ```
   이렇게 수십억 문장에서 다음 단어를 예측하는 과정을 반복하여  
   언어의 통계적 구조를 광범위하게 학습합니다.

2. **미세조정 (Fine-tuning)**  
   - 사전학습된 GPT를 특정 작업(Task)에 맞게 소량의 데이터로 재학습  
        - 예: 대화형 응답, 요약, 번역, 질의응답 등  
   - 여전히 **다음 단어 예측(next token prediction)** 방식을 유지하지만, 학습 데이터의 형태를 **태스크에 맞게 조정**합니다.  
   
        - 예를 들어, "질문–답변(Q&A)" 데이터의 경우 질문과 답변을 분리하지 않고 **하나의 연속된 텍스트 시퀀스로 결합**하여 학습합니다.  
     즉, 모델은 "질문 다음에 어떤 단어가 이어질지(=답변)"를 예측하는 방식으로 훈련됩니다.


   예:  
   ```
   입력 시퀀스:  
   "사용자: Explain quantum computing in simple terms.\n모델: Quantum computing uses qubits that can represent both 0 and 1..."

   → 모델은 위 텍스트 전체를 하나의 문맥으로 보고  
     '모델:' 다음에 이어질 올바른 단어를 순차적으로 예측하며 학습합니다.
   ```

#### 참고 : Attention Mask (Causal Mask)
GPT는 미래 단어를 미리 보면 안 되기 때문에, **Causal Mask(=Triangular Mask)** 를 적용하여  
현재 시점 이전의 단어 정보만 참고하도록 제한합니다.

이로써 GPT는 이전 단어를 기반으로 다음 단어를 생성하는 **자기회귀(auto-regressive)** 언어 모델로 동작하게 됩니다.

#### GPT는 어떻게 문장을 만들어낼까? (Decoding)

- 실제 생성(inference) 시에는 한 번에 문장을 다 뱉는 것이 아니라,  
  **토큰을 하나씩 순차적으로 생성**해 나갑니다.
- 프롬프트(입력 문장)를 넣으면, 모델이 **다음 토큰 분포 $p(x_t \mid x_{<t})$** 를 계산하고  
  여기서 한 토큰을 선택하여 문장 끝에 붙입니다.
- 이렇게 **새로 생성된 토큰까지 포함한 시퀀스를 다시 입력**으로 넣고,  
  다음 토큰 분포를 또 계산하는 과정을 반복하며 문장을 길게 이어 갑니다.

토큰 선택(디코딩) 방식 예시:

- **Greedy decoding**: 가장 확률이 높은 단어만 계속 선택 → 일관성은 좋지만, 다양성이 떨어질 수 있음
- **Sampling (temperature, top-k, top-p)**: 확률 분포에서 **무작위 샘플링**을 하되,  
  상위 k개나 누적 확률 p까지의 후보만 남겨서 샘플링 → 더 다양한, 자연스러운 문장 생성 가능


#### 주요 특징
| 구분 | BERT | GPT |
|------|------|------|
| 구조 | Transformer **Encoder** | Transformer **Decoder** |
| 학습 방식  | MLM + NSP | LM (다음 단어 예측)|
| 방향성 | 양방향(Bidirectional) | 단방향(Left-to-Right) |
| 학습 목표 | Masked LM, NSP | Next Word Prediction |
| 데이터 | 문장 단위 쌍 (마스크 예측용) | 대규모 연속 텍스트 (비지도) |
| 강점 | 문맥 이해, 문장 분류 | 문장 생성, 대화, 서술 |
| 약점 | 생성 불가능 | 문맥 해석 정확도 낮음(초기 모델 기준) |
| 사용 목적     | 이해(분류/추론)      | 생성(Text Generation)

outputs = model.generate(inputs["input_ids"], max_length=30)

# 번역


````python
#  GPT2 기반 한-영 번역 파인튜닝 전체 코드 (완전 정리본)

import torch
from datasets import load_dataset
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    Trainer,
    TrainingArguments,
)

device = "cuda" if torch.cuda.is_available() else "cpu"
print("DEVICE =", device)

# 1. GPT-2 모델 / 토크나이저 로드 + pad/eos 처리
base_model_name = "gpt2"
tokenizer = AutoTokenizer.from_pretrained(base_model_name)

# gpt2는 pad_token이 없음 → eos_token을 대신 pad로 설정
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
    tokenizer.pad_token_id = tokenizer.eos_token_id

model = AutoModelForCausalLM.from_pretrained(base_model_name).to(device)

# 2. 간단 번역 함수 (훈련 전/후 비교용)

def gpt_translate(model, tokenizer, text_ko: str, max_new_tokens: int = 64):
    prompt = (
        "Translate the following Korean sentence into natural English.\n"
        f"Korean: {text_ko}\n"
        "English:"
    )

    input_ids = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=256,
    ).input_ids.to(model.device)

    with torch.no_grad():
        output_ids = model.generate(
            input_ids,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            top_p=0.9,
            temperature=0.7,
            pad_token_id=tokenizer.pad_token_id,
        )

    generated = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    return generated.split("English:")[-1].strip()


# 3. 데이터셋 로드
dataset = load_dataset("strongminsu/ko-en-structured-translations")

SRC_COL = "ko"
TGT_COL = "en"

train_dataset = dataset["train"]
valid_dataset = dataset["test"]   # 검증셋 대체
test_dataset  = dataset["validation"]


# 4. 파인튜닝 전 번역 결과 확인
sample_texts = [
    "이 모델은 한국어를 영어로 번역하는 데 사용됩니다.",
    "오늘 날씨가 정말 좋네요.",
    "내일 회의 일정 다시 한번 확인해 주세요.",
]

print("=== 파인튜닝 전 번역 예시 ===")
for text in sample_texts:
    print(f"[KO] {text}")
    print(f"[EN] {gpt_translate(model, tokenizer, text)}\n")


# 5. 전처리 함수 정의
#    - prompt + target + eos_token
#    - input_ids / labels / attention_mask 반환
max_length = 256

def preprocess_function(example):
    src = example[SRC_COL]
    tgt = example[TGT_COL]

    # ★ ko/en 이 리스트(list[str])로 들어오는 경우 처리 (문제의 핵심)
    if isinstance(src, list):
        src = " ".join(src)
    if isinstance(tgt, list):
        tgt = " ".join(tgt)

    prompt = (
        "Translate the following Korean sentence into natural English.\n"
        f"Korean: {src}\n"
        "English:"
    )

    text = prompt + " " + tgt + tokenizer.eos_token

    encoded = tokenizer(
        text,
        truncation=True,
        max_length=max_length,
        padding=False,
    )

    ids = encoded["input_ids"]

    return {
        "input_ids": ids,
        "labels": ids.copy(),
        "attention_mask": [1] * len(ids),
    }


# 6. 전처리 데이터셋 생성 (중요!)
tokenized_train = train_dataset.map(
    preprocess_function,
    remove_columns=train_dataset.column_names
)

tokenized_valid = valid_dataset.map(
    preprocess_function,
    remove_columns=valid_dataset.column_names
)


# 7. Collator (padding만 수행)
def gpt_collator(features):
    # pad input_ids
    batch = tokenizer.pad(
        {"input_ids": [f["input_ids"] for f in features]},
        padding=True,
        return_tensors="pt"
    )
    
    # pad labels (직접 pad 처리)
    max_len = batch["input_ids"].shape[1]
    labels = []
    for f in features:
        pad_len = max_len - len(f["labels"])
        labels.append(f["labels"] + [-100] * pad_len)

    batch["labels"] = torch.tensor(labels)
    return batch


# 8. Trainer 설정
output_dir = "./gpt2-ko-en-finetuned"

training_args = TrainingArguments(
    output_dir=output_dir,
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    learning_rate=5e-5,
    num_train_epochs=10,
    eval_strategy="epoch",
    logging_steps=50,
    save_strategy="epoch",
    fp16=torch.cuda.is_available(),
    report_to="none",
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_valid,
    data_collator=gpt_collator,
)


# 9. 학습 시작
trainer.train()


# 10. 파인튜닝 후 번역 확인
print("\n=== 파인튜닝 후 번역 예시 ===")
for text in sample_texts:
    print(f"[KO] {text}")
    print(f"[EN] {gpt_translate(model, tokenizer, text)}\n")
````

이 절에서 살펴본 것처럼 최신 사전학습 언어 모델은
  
**BERT 계열**(RoBERTa, DistilBERT, ALBERT 등)을 통해 문맥 이해 성능과 효율성을 높이고,
